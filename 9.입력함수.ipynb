```python
# 두 수를 전달 받아서 두 수의 합을 알려주는 로직
num1 = int(input("첫 번째 정수 입력:\nex)20\n"))
num2 = int(input("두 번째 정수 입력:\nex)20\n"))
result = f"두 수의 합은: {num1 + num2}"

print(result)
```
