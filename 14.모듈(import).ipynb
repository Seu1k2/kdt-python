```python
import math

# 3.14
# .ceil(): 올림
print(math.ceil(3.14))
# .floor(): 내림
print(math.floor(3.99))
# round(): 반올림
print(math.floor(2.5))
# .factorial() == 5!: 5x4x3x2x1
print(math.factorial(5))
# .pi: 원주율
print(math.pi)
```

```python
# .choice(): 시퀀스 자료구조 중에 1개를 선택하여 반환

# 추첨
students = ["김희수", "조수민", "박지호", "혜진", "이상아", "김진휘", "정우영", "오승현", "오슬기", "오지혁", "이예나", "채진성"]
print(random.choice(students))
```

```python
from collections import Counter

# 리스트 또는 시퀀스 개체에서 갯수를 세는 객체

fruits = "apple banana apple banana apple melon".split()
fruits_count = Counter(fruits)
print(fruits_count)
print(fruits_count["apple"])
```

```python
import time

# 시작 시간
start_time = datetime.now()
# .sleep(5): 5초동안 코드가 멈춤
time.sleep(5)

# 종료 시간
end_time = datetime.now()

# 시작 - 종료
elapsed = end_time - start_time

print(f"경과 시간: {elapsed}")
```
