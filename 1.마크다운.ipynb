# 주피터 노트북 단축기
- 코드셀 변경:[ESC] -> [Y]
- 텍스트 셀 변경: [ESC] -> [M]
- 셀 실행: [Ctrl] + [Enter]
- 셀 실행 후 코드셀 생성: [Alt] + [Enter]
- 주석:  [Ctrl] + [/]
- 되돌리기: [Ctrl] + [Z]
- 다시 되돌리기: [Ctrl] + [Shift] + [Z]
- 한줄삭제: [Ctrl] + [D]
- 한줄복사: [Art] + [Shift] + [화살표 위 또는 아래]

# 목차(Header)
# : h1
## : h2
### : h3
#### : h4
##### : h5

# 리스트(list)
1. 장점
   - 간편하고 쉽다
   - 다양한 플랫폼에서 지원한다

2. 단점
    - 표준이 아니다
    - 표현매체마다 다르게 보인다
    - 모든 HTML을 대체하지 못한다



# 용어
- []: 대괄호 (브레이킷)
- {}: 중괄호 (컬리 브레이킷)
- (): 소괄호

## 줄바꿈
- 띄어쓰기 (공백문자) 2개 적으면 줄바꿈
- <br /> 태그를 사용하여 줄바꿈
-  ex) 파이썬 귀도반 로섬이 좋아하는 코미디언의 이름은 <br />몬티 파이썬이다

# 표현식
1. *내용*: 글자 기울이기(italic)
   - 설명문 (Tab키)
   - 파이썬은 퀴도반 로섬이 *좋아하는 코미디언의 이름*을 따서 만든 이름이다.
2. **내용**: 글자 굵게(bold)
    - 파이썬의 로고는 **그리스 로마 신화의 뱀**을 의미하여 생긴 로고이다.
3. ~~내용~~ : 취소선 (line through)
   - ~~C언어~~ 2년걸린 프로젝트를 파이썬으로 한달만에 마무리했다.
4. `내용`: 강조 (back tick)
   - 파이썬의 부모 언어는 `C언어`이다. 즉 파이썬은 c언어에서 파생되었다.

# 태그
- 여는 태그: <태그명>, 태그의 시작을 알리는 태그
- 닫는 태그: </태그명>, 태그의 종료를 알리는 태그
- 내용: <태그명>내용</태그명>, 태그가 표현하는 컨텐츠

- ex) <sup>홍길동</sup>

# 인용문
- 남의 말이나 글을 직접 또는 간접적으로 따온 문장을 인용문장이라고 한다.

> 인용문
>> 중첩 인용문
>>> 중첩 인용문2
>>>> 중첩 인용문3
>>>>> 중첩 인용문4

# 표(table)
- 행과 열로 존재한다. 흔히 표의 현식으로 구성된 것을 테이블이라고 한다.

|001|홍길동|
|002|장보고|
|003|이순신|

# 레이텍(Latex)

- \: 역슬래시
- $\alpha$
- $\sigma$
