## 지역변수
   - 함수 내부에서 선언된 변수이며, 함수에서만 사용이 가능하다.

## 전역변수
   - 함수 밖에서 선언한 변수, 전역에서 사용이 가능하다.

## global문
   - global문을 사용하면 함수 안에서도 전역변수의 값을 수정이 가능하다

```python
# 전역 변수
value = 100

def change_value():
    # 지역 변수
    global value
    value = 20

change_value()

print(value)
```

```python
value = 100

def test():
    print(value)

test()
```

```python
# employee 함수 3개 제작
# def employee1
# def employee2
# def employee3
# 각 함수를 사용하면 employee 회사원이 가지고 있는
# 수입을 받아서 수입을 전역변수에 모두 더하기!

total = 0

def employee1(income):
    global total
    total += income

def employee2(income):
    global total
    total += income

def employee3(income):
    global total
    total += income

employee1(10000)
employee2(20000)
employee3(30000)

print(total)
```
