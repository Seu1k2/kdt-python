```python
# 심화
# 연속되는 글자의 개수 중 가장 많이 반복되는 글자 찾기
# 단, 같은 횟수라면 제일 먼저 들어오는 값을 리턴한다.
# 예상 출력결과
# text = "abcccddeaddddadwefda"
# 제일 많이 반복되는 글자 : d
# 횟수 : 4

text = "abcccddeaddddadwefda"

most_char = text[0]
max_count = 1

current_char = text[0]
current_count = 1

for i in range(1, len(text)):
    if text[i] == current_char:
        current_count += 1
    else: 
        current_char = text[i]
        current_count = 1

    if current_count > max_count:
        most_char = current_char
        max_count = current_count

result = f"가장 연속된 글자가 많은 문자형: {most_char}\n연속 횟수: {max_count}번"
print(result)
```
