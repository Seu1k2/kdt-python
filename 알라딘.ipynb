```python
import requests
from bs4 import BeautifulSoup as bs
import json

url = "https://www.aladin.co.kr/shop/common/wbest.aspx?BranchType=1"

response = requests.get(url)
aladin_soup = bs(response.text, "html.parser")
```

```python
titles = aladin_soup.select(".bo3")
title_list = list(map(lambda tag: tag.text.strip(), titles))

title_list
```

```python
prices = aladin_soup.select(".ss_p2")
price_list = list(map(lambda tag: tag.text.strip(), prices))

price_list
```

```python
stars = aladin_soup.select(".star_score")
star_list = list(map(lambda tag: tag.text.strip(), stars))

star_list
```

```python
print(len(title_list))
print(len(price_list))
print(len(star_list))
```

```python
books = aladin_soup.select(".ss_book_box")
len(books)
```

```python
book_list = []

for book in books:
    title = book.select_one("a.bo3")
    price = book.select_one("span.ss_p2")
    star = book.select_one(".star_score")

    if star == None:
        star_text = "별점 없음"
    else:
        star_text = star.text.strip()

    book_list.append(
        (
            title.text.strip(),
            price.text.strip(),
            star_text
        )
    )

book_list
```
