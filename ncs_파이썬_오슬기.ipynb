```python
# 문장
texts = ["야 너 진짜 바보냐? 왜 그런 실수를 해?", "쟤는 행동하는 게 완전 모지리 같아.", "그런 식의 답변을 내다니 진짜 인공저능아 수준이네."]
# 검열해야 할 단어
badwords = ["바보", "모지리", "인공저능아"]

for text in texts:
    for word in badwords:
        # 문장에 있는 단어를 "****"로 대체
        text = text.replace(word, "****")
    print(text)
```

```python
class Market: # 마켓 클래스
    def __init__(self, name, product, income = 0):
        self.name = name
        self.product = {"name": "토마토", "price": 3000, "stock": 10} # 상품: 토마토, 가격: 3000, 갯수: 10
        self.income = income

    def sell(self, buyer, count): # 판매 함수
        tomato_price = self.product["price"]
        halinyoul = tomato_price * buyer.discount / 100
        halinga = tomato_price - halinyoul
        gageuk = halinga * count
        
        # 유저가 가지고 있는 돈이 상품의 판매 가격보다 < 적다면 판매할 수 없다.
        if buyer.money < gageuk:
            return print("잔액 부족")
            
        # 유저가 구매하는 개수보다 재고의 개수가 > 적다면 판매할 수 없다.
        if self.product["stock"] > count:
            return print("상품 판매 불가")

        # 상품 판매시 유저가 가지고 있는 할인율이 적용되어 판매가 된다.



class Buyer: # 구매자 클래스
     def __init__(self, name, money, discount):
         self.name = name
         self.money = money
         self.discount = discount
```
