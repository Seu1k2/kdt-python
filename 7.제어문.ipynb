```python
num = 12

# 3의 배수인지 확인
if num % 3 == 0:
    print("3의 배수")

if num % 3 == 0:
    print("3의 배수")
```

```python
# if~else문
if num % 3 == 0:
    print("3의 배수")

elif num % 3 == 0:
    print("3의 배수")
```

```python
if num % 3 == 0:
    print("3의 배수")

# else: 위에 있는 모든 조건식을 만족하지 안ㅇㅎ으면
else:
    print("3의 배수가 아닙니다.")
```

```python
# 숫자 2개를 변수에 담고,
# 두 수를 비교해서 큰 수와 작은 수를 출력하기
# 단, 두 수가 같을 수도 있음
# 만약 두 수가 같다면 "두 수는 같습니다"를 출력하기

num1 = 10
num2 = 20

if num1 > num2:
    print("1번")
elif num1 < num2:
    print("2번")
else:
    print("두 수는 같습니다")
```

```python
# 점수 등급
# A: 90, B: 80, C: 70, D: 60, 60미만은 과락

# 사용자에게 이름과 수학, 영어, 국어 평균 점수의 등급을 매기는 로직 구현

# 출력 예시)
# 홍길동님의 총 점은 OOO점, 평균은 OO.OO점으로 OO등급입니다.

name = "홍길동"
math = 92
english = 88
korean = 74

# 총점
total = math + english + korean

# 평균
average = total / 3

# 등급
if average >= 90:
    grade = "A"
elif average >= 80:
    grade = "B"
elif average >= 70:
    grade = "C"
elif average >= 60:
    grade = "D"
else:
    grade = "과락"

# 출력
print(f"{name}님의 총 점은 {total}점, 평균은 {average:.2f}점으로 {grade}등급입니다.")
```
