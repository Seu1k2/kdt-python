## 알라딘 웹 크롤링 데이터 수집 예제
- https://www.aladin.co.kr/shop/common/wbest.aspx?BranchType=1
- "책 제목", "가격", "별점"을 수집하는 코드 작성 후 제출

```python
import requests
from bs4 import BeautifulSoup as bs
import json

url = "https://www.aladin.co.kr/shop/common/wbest.aspx?BranchType=1"

response = requests.get(url)
aladin_soup = bs(response.text, "html.parser")
```

```python
titles = aladin_soup.select("a.bo3")

title_list = list(map(lambda tag: tag.text.strip(), titles))

title_list
```

```python
prices = aladin_soup.select("span.ss_p2")

price_list = list(map(lambda tag: tag.text.strip(), prices))

price_list
```

```python
stars = aladin_soup.select(".star_score")

star_list = list(map(lambda tag: tag.text.strip(), stars))

star_list
```

```python
print(len(title_list))
print(len(price_list))
print(len(star_list))
```

```python
books = aladin_soup.select(".ss_book_box")
len(books)
```

```python
book_list = []

for book in books:
    title = book.select_one("a.bo3")
    price = book.select_one("span.ss_p2")
    star = book.select_one(".star_score")

    if star == None:
        star_text = "별점 없음"
    else:
        star_text = star.text.strip()

    book_list.append(
        (
            title.text.strip(),
            price.text.strip(),
            star_text
        )
    )

book_list
```

# velog 크롤링

```python
options = Options()

user_agent = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/152.0.0.0 Safari/537.36 Edg/152.0.0.0"}
options.add_argument(f"user-agent={user_agent}")

chrome = Chrome(options=options)

velog_url = "https://velog.io/"

chrome.get(velog_url)
time.sleep(5)

recent_btn = chrome.find_elements(By.CLASS_NAME, "HomeTab_tab__viwzb")[1]
recent_btn.click()
time.sleep(5)

for i in range(2):
    chrome.execute_script("window.scrollTo(0, document.body.scrollHeight)")
    time.sleep(5)

velog_soup = (chrome.page_source, "html.parser")
```

#### 실습) velog의 게시글 제목, 내용, 작성자를 각 리스트에 담아서 zip으로 묶기

[("제목", "내용", "작성자"), ...]

```python
titles = velog_soup.select(".PostCard_h4__6dTLA")
title_list = list(map(lambda tag: tag.text, titles))
print(f"제목\n{title_list}")
print()

texts = velog_soup.select(".PostCard_descriptionWrapper__UQSKA")
text_list = list(map(lambda tag: tag.text, texts))
print(f"내용\n{text_list}")
print()

users = velog_soup.select(".PostCard_userInfo__Cu1X5")
user_list = list(map(lambda tag: tag.text, users))
print(f"작성자\n{user_list}")
print()

velog_info = list(zip(title_list, text_list, user_list))
print(f"벨로그리스트\n{velog_info}")
```
