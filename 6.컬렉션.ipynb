# 컬렉션(list, tuple, set, dict)
1. list []
    - 추가, 수정, 삭제 언제나 가능
    - [] 대괄호로 묶어주고 ,(콤마)로 값을 구분한다

2. tuple ()
   - 한 번 생성되고나면 변경이 불가능
   - () 소괄호로 묶어주고 ,(콤마)로 값을 구분한다

3. set {}
   - 순서가 없는 자료 구조(비시퀀스)
   - 중복된 값을 저장이 불가능
   - {} 중괄호로 묶어주고 ,(콤마)로 값을 구분한다

4. dict {key:value}
   - 키 + 값이 한 쌍으로 관리하며 ,(콤마)로 값을 구분한다
   - key는 중복 값을 허용하지 않으며, value는 중복값을 허용한다
   - key를 통해 value를 가져온다

## ※딕셔너리

```python
# 선언
di1 = {}
d2 = set()

print(di1)
print(type(di1))
```

```python
di2 = dict()
print(di2)
print(type(di2))
```

```python
di3 = {
    "a": "apple",
    "b": "banana",
    "c": "cheery",
}

print(di3)
```

```python
# 값을 조회(Read)
# 대괄호 접근법
print(di3["a"])
print(di3.get("a"))
```

```python
# d를 조회
# 대괄호 접근법 없는 키 접근 -> 오류
print(di3["d"])

# .get()으로 없는 키 접근 -> 오류가 나지 않음 
print(di3.get("d", "대체값"))
```

```python
# .get()으로 없는 키 접근 -> 오류
print(di3.get("d", "대체값")) 
```

```python
di3["a"] = "alpha"
print(di3)
```

```python
di3 = {
    "a": "apple",
    "b": "banana",
    "c": "cheery",
    # 수정
    "a": "alpha",
}

print(di3)
```

```python
# .(dot) 접근법
print(di3.items())
print(di3.keys())
print(di3.values())
```
