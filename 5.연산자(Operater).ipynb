## 산술 연산자(+, -, *, /, //, %)
1. +: 더하기 연산자
2. -: 빼기 연산자
3. *: 곱하기 연산자
4. /: 나누기 연산자
5. //: 몫 연산자
6. %: 나머지 연산자
7. ** : 파워연산자, 거듭제곱

    * 시퀀스 연산자(+, *)
    - 순서가 있는 자료구조에 사용할 수 있는 연산자

```python
data1 = 10
data2 = 20

# 정수 + 정수 = 정수
print(data1 + data2)
# 정수 - 정수 = 정수
print(data2 - data1)
# 정수 * 정수 = 정수
print(data2 * data1)
# 정수 / 정수 = 실수(자동 형변환)
print(data2 / data1)
```

```python
#  조건) 
#  나이는 임의로 설정, 삼항 연산자
#  ** 중요 ** 값을 모두 변수에 담아보기

#  실습)
#  1. 나이가 19살 이상이면 "입장 가능" 출력
#  19살 미만이면 "입장 불가" 출력
#  2. 나이가 짝수면 "짝수" 출력
#  나이가 홀수면 "홀수" 출력
#  3. 나이가 16살이면 "당첨" 출력
#  아니면 "꽝입니다" 출력

#  예상 결과)
#  콘솔 결과: [본인이름]님은 [입장 가능]
#  [나이], [홀수], [꽝입니다].  
```

```python
age = 21
name = "홍길동"
is_adult = age >= 19
pass_message = "입장 가능"
not_pass_message = "입장 불가"
even_message = "짝수"
odd_message = "홀수"
lucky_message = "당첨"
unlucky_message = "꽝"

adult_result = pass_message if is_adult else not_pass_message
is_even = age % 2 == 0
even_result = even_message if is_even else odd_message
is_lucky = age == 16
lucky_result = lucky_message if is_lucky else unlucky_message

result = name + "님은 " + adult_result + "\n" + str(age) + ", " + even_result + ", " + lucky_result
print(result)
```

```python
# 탐욕 알고리즘(greedy)
# 사용자의 거스름돈의 최소 동전 개수 구하기
# 동전의 종류 1000, 500, 100, 10
# 12790원

# 예상 출력 결과
# 동전의 개수:
# 1000원: 12개
# 500원: 1개
# 100원: 2개
# 10원: 9개
```

```python
# 변수에 12790을 담음
money = 12790
coin_1000 = 1000
coin_500 = 500
coin_100 = 100
coin_10 = 10

count_1000 = money // coin_1000
money %= coin_1000
count_500 = money // coin_500
money %= coin_500
count_100 = money // coin_100
money %= coin_100
count_10 = money // coin_10
money %= coin_10

result = f"동전의 개수:\n1000원: {count_1000}개\n500원: {count_500}개\n100원: {count_100}개\n10원: {count_10}개"
print(result)
```

```python
#  심화
#  주차 요금 구하기
#  1. 기본 주차 요금은 시간당 1,000원
#  2. 주말이면 총 요금의 20% 추가 요금이 붙는다
#  3. 이용 시간이 5시간 초과이면 총 요금에서 10% 할인
#  4. 최종 요금을 계산하여 출력한다
 
#  주차시간과 주말여부는 임의로 넣는다.
#  모든 값은 변수에 담아서 사용한다.
```

```python
hour = 4
is_weekend = False
total = hour * 1000
total *= 1 + is_weekend * 0.2
total *= 1 - (hour > 5) * 0.1
total = int(total)

print(f"주차 시간: {hour}시간\n주차 비용: {total}원")
```

```python
#  기본
#  1. 키, 나이를 입력 받기
#  2. 놀이공원에서 어린이 놀이기구 탑승 여부 확인하기
#  3. 130cm 이하 그리고 성인은 탑승 불가
#  4. 나머지는 탑승 가능
name = "홍길동"
height = 120
age = 15
pass_message = "탑승 가능"
not_pass_message = "탑승 불가"

enter = height <= 130 and age <= 19
result = pass_message if enter else not_pass_message

print(f"{name}님은 {result}")
```
