```python
# 실습
# 유저(User) 클래스를 생성하고 아래 코드를 완성하시오
# 유저는 이름(name), 돈(돈)를 가지고 있다. 돈은 모두 10000원씩 존재한다
# 유저는 출금(withdraw) 메서드를 통해 유저의 돈을 출금할 수 있다.
# 아래에 코드를 완성하시오.

class User:
    def __init__(self, name, money = 10000):
        self.name = name
        self.money = money

    def info(self):
        print(f"{self.name}님의 잔액: {self.money}원")

    def withdraw(self, withdraw_money):
        # 내가 가진 돈보다 많은 경우 return
        if withdraw_money > self.money:
            print(f"잔액이 부족합니다.\n잔액: {format(self.money, "_")}원")
            return -1
            
        # 출금이 가능한 경우
        # 1. 내가가진 돈을 빼야된다.
        # self.money = self.money - withdraw_money
        self.money -= withdraw_money
    
        # 2. 돈을 반환해야 한다.
        print(f"출금 완료\n잔액:{format(self.money, "_")}")
        return withdraw_money

hong = User("홍길동")
jang = User("장보고")

hong.info()
jang.info()

# 홍길동 출금
hong.withdraw(15000)
```

```python
# 실습, 아래 코드를 설계하시오.
# 1. 사물함(Locker) 클래스가 존재한다.
# 2. 사물함은 주인(owner)과 비밀번호(password), 사물함의 물건(items)가 존재한다.
# 3. 물건을 추가하는 메서드는(add_item) 사물함의 물건에 전달받은 물건을 추가한다.
# 4. 물건을 꺼내는 메서드는(get_item) 사물함의 물건에서 요청받은 물건을 꺼내고 반환한다.
# 5. 사물함은 추가하거나, 꺼낼 때 비밀번호가 일치해야한다.
```

```python
class Locker:
    def __init__(self, owner, password = "0000", items = []):
        self.owner = owner
        self.password = password
        self.items = items

    def add_item(self, item, password):
        # 비밀번호가 틀리면
        if not self.verify_password(password):
            print("비밀번호가 일치하지 않습니다.")
            return
        
        print(f"{item} 추가가 완료되었습니다.")
        self.items.append(item)
        self.show_item()

    def get_item(self, item, password):
        if not self.verify_password(password):
            print("비밀번호가 일치하지 않습니다.")
            return
        
        # 만약 락커 아이템이 없다면
        if not item in self.items:
            print(f"{item}은 목록에 없습니다")
            self.show_item()
            return "찾을 목록을 정확히 입력해주세요."

        # 락커에 아이템이 있는 경우
        print(f"{item}을 찾았습니다")
        remove_item = self.items.pop(self.items.index(item))
        self.show_item()
        return remove_item

    def verify_password(self, view_password):
        return self.password == view_password
    
    def show_item(self):
        print(f"{self.owner}님의 락커 목록\n{", ".join(self.items)}")

hong_locker = Locker("홍길동", "1234", ["모자", "지갑", "노트북", "마우스"])

hong_locker.show_item()
hong_locker.add_item("키보드", "1234")
```

```python
# 실습, 아래 코드를 설계하시오
# 1. 학교(Student) 클래스가 존재한다.
# 2. 학교에는 이름(name) 학생들(students)이 존재한다.
# 3. 학생들은 각각 이름, 나이, 성별, 국어점수, 영어점수, 수학점수를 가지고 있는 딕셔너리(dict) 객체이다
# 4. 학교 클래스는 학생들의 총 인원수를 확인 할 수 있는 함수(get_total_count)가 존재한다.
# 5. 학교 클래스는 학생들의 총점으로 등수를 출력해 줄 수 있는 메서드(print_student_rank)가 존재한다.
# - 총점이 높은 학생들로 내림차순 정렬된다.`
```

```python
class Student:
    def __init__(self, name, students=[]):
        self.name = name
        self.students = students

    # 총 학생 수
    def get_total_count(self):
        return len(self.students)

    # 학생 총점으로 등수 출력
    def print_student_rank(self):
        # 총점 기준 내림차순 정렬
        rank_students = sorted(
            self.students,
            key=lambda student: student["korean"] + student["english"] + student["math"],
            reverse=True
        )

        # 등수 출력
        for i, student in enumerate(rank_students):
            total = student["korean"] + student["english"] + student["math"]

            print(f"{i + 1}등: {student['name']} - 총점: {total}점")


students = [
    {
        "name": "홍길동",
        "age": 20,
        "gender": "남",
        "korean": 80,
        "english": 90,
        "math": 70
    },
    {
        "name": "김철수",
        "age": 21,
        "gender": "남",
        "korean": 90,
        "english": 95,
        "math": 100
    },
    {
        "name": "이영희",
        "age": 19,
        "gender": "여",
        "korean": 100,
        "english": 80,
        "math": 85
    }
]


school = Student("파이썬학교", students)

print(f"총 학생 수: {school.get_total_count()}명")
school.print_student_rank()
```

```python
class Store:
    def __init__(self, name, products = []):
        self.name = name 
        self.products = products

    def sell(self, product):
        # 상품이 없다면, 판매하면 안된다.
        if not product in self.products:
            print(f"{product} 상품을 판매할 수 없습니다")
            self.show_products()
            return 

        # 상품이 있다면
        print(f"{product} 판매 완료")
        sell_item = self.products.pop(self.products.index(product))
        self.show_products()
        return sell_item

    def insert_product(self, product):
        if product in self.products:
            print(f"이미 상품이 존재합니다")
            return 
        self.products.append(product)
        print(f"{product} 상품 추가 완료")
        self.show_products()

    def insert_products(self, *args):
        # self.products = self.products + list(args)
        self.products += list(set(args))
        print(f"{", ".join(args)} 상품 추가 완료")
        self.show_products()

    def show_products(self):
        print("현재 상품 목록:\n" + ", ".join(self.products))

market = Store("중고나라", ["아이폰", "헤드셋", "신발", "옷"])
print(market.sell("아이폰"))
market.insert_product("모자")
market.insert_products("모자", "핸드폰", "가방", "모자", "모자", "모자")
```

```python
# 기초 풀이
# 실습 아래코드를 설계하시오.
# 1. 상점(Store) 클래스가 존재한다.
# 2. 상점은 이름(name), 상품들(products)를 가지고 있다.
# 3. 상점은 상품 중 구매할 상품을 전달받아서 판매하는 (sell)함수를 존재한다.
# 4. 상점에는 상품을 추가할 수 있는 함수(insert_product)가 존재한다.
# 5. 상점에는 상품들을 여러개 한 번에 추가할 수 있는 함수(insert_products)가 존재한다.
```

```python
class Store:
    def __init__(self, name, products=[]):
        self.name = name
        self.products = products

    # 상품 하나 추가
    def insert_product(self, product):
        self.products.append(product)
        print(f"{product} 상품 추가가 완료되었습니다.")
        self.show_products()

    # 상품 여러 개 추가
    def insert_products(self, products):
        for product in products:
            self.products.append(product)

        print("상품들이 추가되었습니다.")
        self.show_products()

    # 상품 판매
    def sell(self, product):
        # 상품이 상점에 없는 경우
        if product not in self.products:
            print(f"{product}은(는) 없는 상품입니다.")
            self.show_products()
            return "상품명을 정확히 입력해주세요."

        # 상품이 상점에 있는 경우
        print(f"{product}을(를) 판매했습니다.")

        sell_product = self.products.pop(
            self.products.index(product)
        )

        self.show_products()

        return sell_product

    # 현재 상품 목록 출력
    def show_products(self):
        print(f"{self.name}의 상품 목록\n{', '.join(self.products)}")


store = Store(
    "편의점",
    ["콜라", "사이다", "과자", "라면"]
)

store.show_products()

# 상품 하나 추가
store.insert_product("초콜릿")

# 상품 여러 개 추가
store.insert_products(["우유", "물", "커피"])

# 상품 판매
store.sell("콜라")
```

```python
class School:
    def __init__(self, name, students = []):
        self.name = name
        self.students = [{**student, "total": student["kor"] + student["math"] + student["eng"]} for student in students]
        self.student_count = len(students)
    
    def add_student(self, student):
        self.student_count += 1
        self.students.append({**student, "total": student["kor"] + student["math"] + student["eng"]})
        print("학교 등록 완료!")

    def remove_student(self, student):
        removed_student = self.students.pop(self.students.index({**student, "total": student["kor"] + student["math"] + student["eng"]}))
        print(f"{removed_student["name"]} 퇴학")
    
    def get_total_count(self):
        print(f"학교의 총 인원수: {self.student_count}명")
        for student in self.students:
            print("- " + student["name"])

    def print_student_rank(self):
        ranked_list = sorted(self.students, key=lambda student: student["total"], reverse=True)
        for idx, student in enumerate(ranked_list):
            print(f"{idx + 1}등 {student["name"]}님, 총 점: {student["total"]}점")

students = [
    {"name": "홍길동", "age": 20, "gender": "남", "kor": 50, "eng": 65, "math": 80},
    {"name": "장보고", "age": 30, "gender": "여", "kor": 80, "eng": 42, "math": 40},
    {"name": "이순신", "age": 35, "gender": "남", "kor": 70, "eng": 38, "math": 77},
]

kcs = {"name": "김철수", "age": 25, "gender": "남", "kor": 20, "eng": 4, "math": 76}
kyh = {"name": "김영희", "age": 27, "gender": "여", "kor": 100, "eng": 88, "math": 100}

korea_ai = School("코리아AI아카데미", students)
korea_ai.add_student(kcs)
korea_ai.add_student(kyh)
# korea_ai.get_total_count()
korea_ai.print_student_rank()
korea_ai.remove_student(kcs)
```

```python
# 동물과 고양이의 상속관계를 파악해서 상속시키고,
# 동물 클래스에는 먹기 메서드를 구현하고,
# 고양이 클래스에서는 할퀴기 메서드를 구현.

class Animal:
    def __init__(self, name):
        self.name = name

class Cat(Animal):
    def __init__(self, name, eat, scratch):
        super().__init__(name)
        self.eat = eat
        self.scratch = scratch
```

```python
class Animal:   # 부모 클래스
    def __init__(self, name, age):
        self.name = name
        self.age = age
        print(f"이름: {name}\n나이: {age}")

class Cat(Animal):   # 자식 클래스
    def __init__(self, name, age):   
        super().__init__(name, age)   # 부모 클래스에서 name, age 물려 받음

    def action(self, action = "할퀴기"):   # "할퀴기" 함수 생성
        print(f"{self.name}이/가 {action} 했다")
    
cat = Cat("나비", 5)   # 부모 클래스에서 작성한 print문을 물려받아서 부모클래스에 적용 해둔 출력값 나옴!
cat.action()   # "할퀴기" 함수 사용
```

```python
# 동물, 사람, 원숭이
class Animal:
    def __init__(self, name):
        self.name = name

    def eat(self):
        print("먹기")

    def sleep(self):
        print("자기")

    def walk(self):
        print("두 발 걷기")

class Person(Animal):
    # 사람은 직업
    def __init__(self, name, job):
        super().__init__(name)
        self.job = job
    
class Monkey(Animal):
    # 원숭이는 바나나
    def __init__(self, name, banana):
        super().__init__(name)
        self.banana = banana

    def monkey_walk(self):
        print("원숭이는 네 발로 걷기")
```

```python
hong = Person("홍길동", "개발자")
monkey = Monkey("김숭이", "노란 바나나")

hong.eat()
monkey.eat()

hong.walk()
monkey.walk()
monkey.monkey_walk()
```

```python
class Snake(Animal):
    # 사람은 직업
    def __init__(self, name, poison):
        super().__init__(name)
        self.poison = poison

    def snake_walk(self):
        print("기어다님")

snake = Snake("방울뱀", "없음")
snake.walk() # 뱀도 walk를 못 씀
snake.snake_walk()
```

```python
# Phone클래스를 상속
# 이름(name), 가격(price)
# 전화기능(call), 인터넷(browse)

# Phone2G(Phone), Phone3G(Phone2G), Phone4G(Phone3G)
```

```python
class Phone:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def call(self):
        print("전화하는 기능")

    def browse(self):
        print("인터넷")
```

```python
# Phone2G(Phone), Phone3G(Phone2G), Phone4G(Phone3G)
class Phone2G(Phone):
    # name, price, call, browse
    def __init__(self, name, price):
        super().__init__(name, price)

    # 오버라이딩(재정의)
    def browse(self):
        print("2G 인터넷")

    # 2G의 +@ == 자식만의 코드
    def sms(self):
        print("문자 발송")
```

```python
class Phone3G(Phone2G):
    def __init__(self, name, price, brand):
        super().__init__(name, price)
        # 자식만의 코드
        self.brand = brand

    # 재정의(오버라이딩)
    def browse(self):
        print("3G 인터넷")

    # 자식만의 코드 + @
    def stream_vidio(self):
        print("영상통화")
```

```python
# Phone4G(Phone3G)
# AI 탑재 기능, 인터넷 무료

class Phone4G(Phone3G):
    def __init__(self, name, price, brand, lte):
        super().__init__(name, price, brand)
         # 자식만의 코드
        self.lte = lte

    # 재정의(오버라이딩)
    def browse(self):
        print("4G 인터넷")
    # 자식만의 코드 + @
    def ai(self):
        print("AI 기능 탑재")

    def free(self):
        print("free 인터넷")
```

```python
phone = Phone("핸드폰", 10000)
phone.browse()
phone.call()

phone2G = Phone2G("2G", 100000)
phone2G.browse()
phone2G.call()
phone2G.sms()

phone3G = Phone3G("3G", 1000000, "삼성")
phone3G.browse()
phone3G.call()
phone3G.sms()
phone3G.stream_vidio()

phone4G = Phone4G("4G", 3000000, "삼성", "lte")
phone4G.browse()
phone4G.call()
phone4G.sms()
phone4G.stream_vidio()
phone4G.ai()
phone4G.free()
```

```python
class Knight:
    def __init__(self, name, hp):
        self.name = name
        self.hp = hp

    def attack(self):
        print("검으로 공격한다.")

class ArchMage:
    def __init__(self, name, hp, mp):
        self.name = name
        self.hp = hp
        self.mp = mp

    def magic(self):
        print("마법으로 공격한다.")

class User(Knight, ArchMage):
    def __init__(self, name, hp, mp):
        Knight.__init__(self, name, hp)
        ArchMage.__init__(self, name, hp, mp)

user = User("홍길동", 100, 50)
print(user.name)
print(user.hp)
print(user.mp)
```

# 객체간의 프로그래밍

```python
# 은행, 유저, 송금, 출금
class Bank:
    def __init__(self, name, bank_money = 1000000):
        self.name = name
        self.bank_money = bank_money

    # 입금
    def deposit(self, user, money):
        # 은행에도 1만원 입금
        self.bank_money += money
        # user돈도 1만원 출금
        user.user_money -= money
        
    # 출금
    def withdraw(self, user, money):
        # 은행에 돈만큼 -
        self.bank_money -= money
        # 유저에게 돈만큼 +
        user.user_money += money

class User:
    def __init__(self, name, user_money = 0):
        self.name = name
        self.user_money = user_money

kb = Bank("국민은행")
jh = User("지효", 100000)
```

```python
 kb.deposit(jh, 10000)
print(jh.user_money)

kb.withdraw(jh, 50000)
print(jh.user_money)
```

```python
# 쿠폰 할인율
# 상점(Store), 고객(Customer)
# 판매(sell) 
# 상품: 모자 - 10000

class Store:
    def __init__(self, name):
        self.name = name
        self.product = {"name": "모자", "price": 10000}

    def sell(self, customer, count):
        price = self.product["price"]
        discounted_price = price - int(price * (customer.discount / 100))
        customer.money -= price * count
        print(f"원래 상품의 가격: {price}원, 할인된 가격: {discounted_price}원")
        
class Customer:
    def __init__(self, name, money, discount):
        self.name = name
        self.money = money
        self.discount = discount
```

```python
store = Store("인터넷마켓")
hong = Customer("홍길동", 50000, 15)
jang = Customer("장보고", 100000, 80)

store.sell(hong, 1)
store.sell(jang, 1)
```

```python
# 마켓 문제
# - 마켓(Market) 클래스
# 1) 마켓은 이름(name), 판매 상품(product), 수입(income)의 필드를 갖는다.
class Market:
    def __init__(self, name, product, income = 0):
        # product -> dict
        self.name = name
        self.product = product 
        self.income = income

    # 3) 마켓은 유저(모름)에게 상품을 판매할 수 있는 함수(sell)을 가지고 있다.
    # - 유저가 구매하는 개수보다 재고의 개수가 적다면 판매할 수 없다.
    # - 상품 판매시 유저가 가지고 있는 할인율이 적용되어 판매가 된다.
    
    # ex) 유저의 할인율이 10%라면 2700원이 결제된다.
    # - 판매된 상품의 모든 수익금은 income에 누적된다.
    def sell(self, buyer, count):
        # 1. 할인율을 들고와서 할인된 가격을 먼저 구한다.
        tomato_price = self.product["price"]
        tomato_stock = self.product["stock"]
        money = buyer.money
        
        discount_price = int(tomato_price * (100 - buyer.discount) / 100)
        total_price = discount_price * count

        # - 유저가 가지고 있는 돈이 상품의 판매 가격보다 적다면 판매할 수 없다.
        if money < total_price:
            return print(f"""
                상품을 구매할 수 없습니다.\n
                총 구매 가격: {total_price}원\n
                {buyer.name}님의 잔액: {money}원
            """)

        # - 유저가 구매하는 개수보다 재고의 개수가 적다면 판매할 수 없다.
        if tomato_stock < count:
            return print(f"""
                토마토의 재고가 부족합니다. 현재 재고: {tomato_stock}개\n
                구매하려는 개수: {count}개
            """)

        # 사용자가 구매할 수 있을 때
        # - 상품의 재고를 사용자가 구매한 만큼 빼고 수정한다.
        # -> 같은 key로 재할당
        self.product["stock"] = tomato_stock - count
        # self.product["stock"] = {**self.product, "stock": tomato_stock - count}
        
        # - 사용자의 돈도 총 구매 가격만큼 빼야한다.
        money = money - total_price
        
        # - 마켓의 수익도 추가해야한다.
        self.income = self.income + total_price

        print(f"""
            {self.name}을 이용해주셔서 감사합니다.😎\n
            구매한 상품: {self.product["name"]}\n
            구매한 개수: {count}\n
            총 구매한 가격: {format(total_price, "_")}원
        """)
        
# 2) 판매 상품은 다음과 같다
# - 이름(name)은 “토마토”이며, 토마토의 가격(price)은 3000원, 재고(stock)은 10개이다.
tomato = {"name": "토마토", "price": 3000, "stock": 10}
market = Market("희희마켓", tomato)

# - 구매자(Buyer) 클래스
# 1) 구매자는 이름(name), 돈(돈), 할인율(discount)의 필드를 갖는다.
class Buyer:
    def __init__(self, name, money, discount):
        self.name = name
        self.money = money
        self.discount = discount

# 2) 총 3명의 Buyer를 각각 객체화
# - 홍길동, 15000, 5
# - 장보고, 10000, 20
# - 이순신, 30000, 50
hgd = Buyer("홍길동", 15000, 5)
jbg = Buyer("장보고", 10000, 20)
lss = Buyer("이순신", 30000, 50)

# 구매자의 할인율을 적용받아 마켓에서 토마토를 구매를 아래와 같이 구매하였다.
# 마켓의 총 수익과 각 구매자의 잔액을 출력하시오.
# - 홍길동 6개 구매
market.sell(hgd, 6)

# - 장보고 2개 구매
market.sell(jbg, 2)

# - 이순신 8개 구매
market.sell(lss, 8)
```
