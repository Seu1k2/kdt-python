# 콜백함수

```python
# 두 수를 더해서 출력하는 함수
# 매개변수 3개 (num1, num2는 변수 / callback은 함수)
def sum1(num1, num2, callback):
    # 함수 사용
    callback(num1 + num2)

sum1(10, 20, print)
```

```python
# 성과 이름을 붙이고 님을 붙여서 반환
def get_full_name(last_name, first_name, callback):
   return callback(last_name + first_name)

def add_nim(name):
    return name + "님"

get_full_name("홍", "길동", add_nim)
```

```python
# 이름을 전달하면 이름을 5번 출력해주는 함수(콜백함수 사용)
def print_name(name):
    print(name)

def repeat(name, callback):
    for i in range(5):
        callback(name)

repeat("철수", print_name)
```

```python
# 이름을 전달하면 이름을 5번 출력해주는 함수(콜백함수 사용)
def print_name(name):
    print(name)

def repeat(name, callback):
    for i in range(5):
        callback(name)

repeat("철수", lambda name: print(name))
```

```python
# 두 수를 더하고 2배 곱한 결과를 반환
def add(num1, num2, callback):
    return callback(num1 + num2)

def get_square(num):
    return num ** 2

add(10, 20, lambda num: num ** 2)
```

```python
# 두 수를 더하고 2배를 곱한 결과에 10배를 만들고 반환
def add(num1, num2, callback):
    return callback(num1 + num2)

def get_square(num, callback):
    return callback(num ** 2)

def get_multiple_by_10(num):
    return num * 10

add(10, 20, lambda add_result: get_square(add_result, get_multiple_by_10))
```

```python
add(10, 20, lambda add_result: get_square(add_result, lambda result: result * 10))
```

# 람다 실습

```python
# 두 정수의 결과를 곱하고 출력
# 횟수를 전달하면 횟수까지의 합을 반환
# 1~100까지 짝수만 더한 뒤 결과 값의 2를 곱한 후 출력
# 두 수를 더하고 홀수인지 짝수인지 비교한 후 출력
```

```python
# 두 정수의 결과를 곱하고 출력
def multiful(num1, num2, callback):
    callback(num1 * num2)

multiful(10, 20, lambda num: print(num))
```

```python
# 횟수를 전달하면 횟수까지의 합을 반환
def get_total(count, callback):
    callback(count)

def repeat(count):
    total = 0
    for i in range(count):
        total += i + 1
    callback(total)

get_total(10, lambda total: print(total))
```

```python
# 1~100까지 짝수만 더한 뒤 결과 값의 2를 곱한 후 출력
def get_total_even(callback):
    callback(sum([i for i in range(0, 101, 2)]))
    
def multiple(num):
    print(num * 2)

get_total_even(multiple)
```

```python
# 1~100까지 짝수만 더한 뒤 결과 값의 2를 곱한 후 출력
def get_total_even(callback):
    callback(sum([i for i in range(0, 101, 2)]))
    
def multiple(num, callback):
    callback(num * 2)

get_total_even(lambda total: multiple(total, print))
```

```python
# 분리
def add(num1, num2, callback):
    callback(num1 + num2)

def confirm_odd_and_even(num, callback):
    callback("짝수" if num % 2 == 0 else "홀수")

add(10, 20, lambda x : confirm_odd_and_even(x, print))
```

# filter, map, reduce


- interable한 객체들을 순차적으로 순회하여 결과를 변환하고 반환(리턴)하는 함수

```python
li1 = [(i + 1) * 10 for i in range(10)]
li1
```

```python
for idx, value in enumerate(li1):
    li1 [idx] = value * 10

print(li1)
```

```python
li2 = [(i + 1) * 10 for i in range(10)]
li2
```

## 1. map

```python
# map(li2)
list(map(lambda num: num * 10, li2))
```

```python
li3 = [(i + 1) for i in range(50)]
print(li3)
```

```python
li4 = list(map(lambda num: num % 2 == 1, li3))

print(li3)
print(li4)
```

## 2. filter 함수

```python
# filter
# 짝수만 걸러내기

print(list(list(filter(lambda num: num % 2 == 0, li3))))
```

```python
li5 = [chr(ord('a') + i) for i in range(26)]
print(li5)
```

```python
# 짝수 인덱스의 값들만 걸러내고 싶을 때
# acegik,....
li6 = list(filter(lambda tu: tu[0] % 2 == 0, enumerate(li5)))
print(li6)
```

```python
li7 = list(map(lambda c: c[1], li6))
print(li7)
```

```python
# 1~10까지 리스트에 담고, 3배를 곱한 리스트로 변경
li8 = [i for i in range(10)]
li9 = list(map(lambda num: num * 3, li8))
li9

# 100~1까지 리스트에 담고, 5의배수만 걸러낸 리스트로 변경
li10 = [100 - i for i in range(100)]
li11 = list(filter(lambda num: num % 5 == 0, li10))
print(li11)
```

## 3. reduce 함수
- iterable한 객체들을 순차적으로 순회하여 조건식으로 만족하는 값들만 모아서 반환해주는 함수

```python
from functools import reduce

# 1부터 10까지 모든 합을 구하기
li11 = [ i + 1 for i in range(10)]
li11

reduce(lambda acc, crr: acc + crr, li11)
```

```python
# 1~10까지 리스트에 담고 값이 5보다 크다면 5배로 변경
# 1~30까지 배열에 담고 3을 곱한 뒤 출력
# 소문자 알파벳들을 담고 아래와 같이 변경 후 출력
# aBcDeF...
```

```python
li1 = [i + 1 for i in range(10)]
li2 = list(map(lambda num: num * 5 if num > 5 else num, li1))

print(li2)
```

```python
li5 = [chr(ord('a') + i) for i in range(26)]
print(li5)

li6 = list(map(lambda c: chr(ord(c) - 32) if ord(c) % 2 == 0 else c, li5))
print(li6)
print("".join(li6))
```
