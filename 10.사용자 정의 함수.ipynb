```python
# 두 수를 전달하면 두 수의 큰 값과 작은 값을 비교하여 반환
def get_min_and_max(num1, num2):
    if num1 == num2:
        return "두 수가 같습니다"

    # if num1 > num2:
    #    return f"큰 수: {num1}, 작은 수{num2}"
    # else:
    #    f"큰 수: {num2}, 작은 수{num1}"

    return f"큰 수: {num1}, 작은 수{num2}" if num1 > num2 else f"큰 수: {num2}, 작은 수{num1}"
```

```python
# 두 수를 전달하면 두 수의 큰 값을 반환해주는 함수
def get_max(num1, num2):
    if num1 == num2:
        return -1

    return num1 if num1 > num2 else num2
```

```python
# 2) 어떤 값을 전달하면 그 값까지 모두 더한 값을 출력해주는 함수
def add(num):
    total = 0
    for i in range(num):
        total += i + 1
    print(total)

add(15)
```

```python
# 3) 어떤 값을 전달하면 홀수인지 짝수인지 확인해서 결과를 반환해주는 함수
def get_odd_and_even(num):
    return "짝수" if num % 2 == 0 else "홀수"

get_odd_and_even(11)
```

```python
# 4) 어떤 값을 전달하면 실수인지 정수인지 확인해주는 함수
def get_integer_or_float(num):
    if (int(num) == num) and (not "." in str(num)):
        return "정수"

    return "실수"

get_integer_or_float(num)
```

```python
# 5) 문자열을 전달하면 찾을 글자의 개수를 반환해주는 함수
# ex) my_func("abbacad", "a")
# 3 

def get_count(text, c):
    count = 0
    for i in text:
        if i == c:
            count += 1

    return count

get_count("apple", "p")
```

```python
# 6) 문자열을 전달하면 글자를 뒤집어서 반환해주는 함수
# ex) "abcdef" -> "fedcba"

def reverse(text):
    return text[::-1]

reverse("abcdef")
```

```python
# 7) 0, 1을 마구잡이로 5개 전달했을 때
# 0의 개수와 1의 개수를 반환해주는 함수
# ex) my_func(0, 0, 1, 1, 0)
# 0의 개수 3개, 1의 개수 2개
def get_count_zero_and_one(num1, num2, num3, num4, num5):
    ones = 0
    
    if num1:
        ones += 1
    if num2:
        ones += 1
    if num3:
        ones += 1
    if num4:
        ones += 1
    if num5:
        ones += 1
    zeros = 5 - ones
    return f"1의 개수: {ones}, 0의 개수: {zeros}"
```

```python
# 몇 개를 전달하더라도 가장 큰 값과 가장 작은 값을 반환해주는 함수
# ex) my_func(10, 100, 15, 20, 25, 10)

# 가장 큰 값: 100, 가장 작은 값: 10


# 심화) 10개이상으로 전달하면 "10개 이하만 가능합니다."
# 심화) count=7를 전달하면 7개까지만 받을 수 있도록 처리하기
```

```python
def get_min_and_max(*args):
    min_num = args[0]
    max_num = args[0]

    for i in args[1:]:
        if i < min_num:
            min_num = i
        
        if i > max_num:
            max_num = i
    return f"가장 큰 값: {max_num}, 가장 작은 값: {min_num}"

get_min_and_max(10, 100, 15, 20, 27, 5)
```
