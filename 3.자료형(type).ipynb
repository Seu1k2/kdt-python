# 기본 자료형(Data type)
   - 숫자 자료형      int      양의 정수, 음의정수, 0   0, 1, 100, ...
   - 숫자 자료형      float      소수점이 있는 수      0.0, 3.14, ...
   - 논리 자료형      bool      참 또는 거짓      True, False, 값이 있을 때
   - 문자열 자료형      str      문자열         '', "", ...

# 형변환(casting)
- 타입을 다른 타입으로 변환시키는 것

### 자동 형변환

```python
# 정수 + 실수 = 실수
10 + 10.05
```

```python
10 + 0.
```

```python
# 문자열 + 정수 = 연산 X
"10" + 10
```

```python
# 논리 + 정수 = 정수
True + 10
```

```python
# 논리 - 정수 = 정수
False - 5
```

### 강제 형변환
- 타입을 변경하고 싶으면 변경할 타입 이름 뒤에 소괄호()를 감싼 함수에게 값을 넘긴다.
- ex) int("10"): 문자열을 정수로 변경

```python
float(10)
```

```python
bool(1)
```

```python
bool(1)
print(bool(""))
print(bool(0))
print(bool([]))
print(bool({}))
print(bool(10))
print(bool(100))
print(bool(10.5))
print(bool(True))
```

```python
# 반올림은 X
print(int(10.5))
print(int(10.9))
```
