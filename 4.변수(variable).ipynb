# 변수 선언 시 주의사항
1. 되도록 선언부에 한꺼번에 선언한다. (영역 상단)
2. 초기화를 해준다.

## 변수명 선언 시 주의사항
1. 문자로 시작해야 한다.
2. 특수문자는 사용할 수 없다. (단, _는 허용한다.)
3. 소문자로 시작한다.
4. 공백을 사용할 수 없다.
5. 명사로 사용한다.
6. 되도록 한글은 사용하지 않는다.
7. 뜻이 있는 단어를 사용한다.

ex)
- a, b, c, d, e, ... (X)
- data, number, age, name, ... (O)

```python
# 다음 값들을 알맞는 이름의 변수에 담고
# 정수형으로 출력하기
# 형변환 + 변수
# "a", False, "5", "10.7"
```

```python
# 알고리즘
# 1. 타입을 확인한다. ("a", False, "5", "10.7")
# "a": 문자열(str)
# False: 논리형(bool)
# "5": 문자열(str)
# "10.7": 문자열(str)

# 2. 변수의 이름을 정한다. (data1, data2, data3, data4, ???)
# 3. 값들을 하나씩 변수에 담는다
data1 = "a"
data2 = False
data3 = "5"
data4 = "10.7"

# 4. 어떻게 하면 정수형으로 바꿀 수 있을까? 강제 형변환
# int(): 정수형으로 바꿔주는 함수

# 5. int를 사용해서 바꾼다. 변수에 있는 값을 넘겨줘야겠다!
# int(data1)

# 5-1. "a"를 숫자형으로 바꿀 수 있는 방법이 없을까?
# bool("a") -> True 값이 있다는 것을 표현하기때문에 1을 표현하기도 하는구나!

# 5-2. int(data1) 직접사용하면 안된다.
int(bool(data1))

# 6. 결과를 출력한다. -> print()함수를 사용하면 출력할 수 있다.
print(int(bool(data1)))
```

```python
# int(data4)
# print(data4)

# 문자형 -> 실수형 -> 정수형
# 문자형 -> 정수형
int(float(data4))
```
