# with문
- resource를 사용한 뒤 자동으로 반납하여 안전하게 관리해주는 문법

```python
with open(file_path, "w", encoding="utf-8") as f:
    f.write("안녕!")
    f.write("하세요!")
    # f.close() 모든 행이 실행되고 마지막에 close()가 자동으로 실행
```

# 파일 수정

```python
foods = ["돈까스", "제육볶음", "김치찌개", "순두부찌개", "된장찌개", "순대국", "센베이"]

file_name = "food_list.txt"
file_path = os.path.join(base_path, file_name)

with open(file_path, "w", encoding="utf-8") as f:
    f.writelines([food + "\n" for food in foods])
```

```python
# 1. 읽어오기
with open(file_path, "r", encoding="utf-8") as f:
    foods = f.readlines()
    foods[6] = "삼겹살\n"

print(foods)

with open(file_path, "w", encoding="utf-8") as f:
    f.writelines(foods)
    print(f"{file_name}파일 수정이 완료되었습니다.")
```

```python
# 돈까스, 된짱찌개 삭제
with open(file_path, "r", encoding="utf-8") as f:
    foods = f.readlines()
    for food in foods:
        if food == "돈까스\n" or food == "된장찌개\n":
            foods.remove(food)
            
with open(file_path, "w", encoding="utf-8") as f:
    f.writelines(foods)
    print(f"{file_name}이 수정되었습니다.")
```

```python
# 돈까스, 된짱찌개 삭제
with open(file_path, "r", encoding="utf-8") as f:
    foods = f.readlines()
    new_foods = []
    for food in foods:
        if food == "돈까스\n" or food == "된장찌개\n":
            continue
        new_foods.append(food)

new_foods
```

```python
# CSV(Comma Separated Value)
import csv

file_name = "동물미용실.csv"
file_path = os.path.join(base_path, file_name)

with open(file_path, "r", encoding="cp949") as f:
    reader = csv.reader(f)

    header = next(reader)
    for row in list(reader)[:5]:
        print(row)
```

```python
file_name = "할일목록.csv"
file_path = os.path.join(base_path, file_name)

with open(file_path, "w", encoding="utf-8-sig", newline="") as f:
    writer = csv.writer(f)
    # 첫 줄 제목(열)
    writer.writerow(["날짜", "할일 목록"])
    writer.writerows([("2026-08-28", "퇴실"), ("2026-08-28", "복습"), ("2026-08-28", "잠자기")])

    print(f"{file_name}이 생성되었습니다")
```
